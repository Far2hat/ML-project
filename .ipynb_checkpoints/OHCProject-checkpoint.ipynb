# DATA PREPROCESSING # 

```python
#Remove duplicate questions (keep the first occurrence)
df = df.drop_duplicates(subset="Question", keep="first")
```

```python
#Trim whitespace and remove question marks from the 'Question' column
df["Question"] = df["Question"].str.strip().str.replace("?", "", regex=False)

#Strip whitespace from 'Answer' column as well
df["Answer"] = df["Answer"].str.strip()
```

```python
#Capitalize the first letter of each question and answer
df["Question"] = df["Question"].str.capitalize()
df["Answer"] = df["Answer"].str.capitalize()
print(df.head())
```

```python
 #Preprocessing Function (for NLP)
stop_words = set(stopwords.words("english"))
```

```python
def preprocess_text(text):
    text = text.lower()  # lowercase
    text = re.sub(r"[^a-zA-Z0-9\s]", "", text)  # remove punctuation
    words = word_tokenize(text)  # tokenize
    words = [word for word in words if word not in stop_words]  # remove stopwords
    return " ".join(words)  # rejoin
```

```python
#  Apply Preprocessing to Question Column to improve keyword search and others
df["Preprocessed_Question"] = df["Question"].apply(preprocess_text)

# Save the Cleaned and Preprocessed Dataset
df.to_csv("Dataset2.csv", index=False)

# Preview
df[["Question", "Preprocessed_Question", "Answer"]].head()
```

# MODEL BUILDING #

```python
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer
import numpy as np
```

```python
#This imports Python’s built-in **os** module,\
#which lets you interact with the *operating system* \
#like setting environment variables, working with files and directories.
import os
os.environ["HF_HUB_DISABLE_SYMLINKS_WARNING"] = "1"
```

###  Load a pretrained model for embeddings###

```python
#Load a smart model that turns sentences into numbers so you can search and compare based on meaning, not just exact words.
model = SentenceTransformer('all-MiniLM-L6-v2')  # Fast & effective for Q&A
```

### Create embeddings ###

```python
#turning every cleaned question into a vector of numbers using a pretrained smart model, so you can later compare and search based on meaning.
embeddings = model.encode(df["Preprocessed_Question"].tolist())
```

### Compute embeddings for all questions ###

```python
#encoding all your cleaned questions into fast, deep-learning-friendly tensors for quicker and smarter searching.
question_embeddings = model.encode(df["Preprocessed_Question"].tolist(), convert_to_tensor=True)
```

### Save FAISS index ###

```python
# creating a fast search engine for your question embeddings using FAISS(Facebook AI similarity search
index = faiss.IndexFlatL2(embeddings.shape[1])
index.add(np.array(embeddings))
```

### Save data for reuse ###

```python
#saving all your important work (dataset, embeddings, search engine) into files, so you can reuse them instantly later without rebuilding everything.
with open("faq_data.pkl", "wb") as f:
    pickle.dump({"df": df, "embeddings": embeddings}, f)
faiss.write_index(index, "faq_index.index")
```

```python
# Sample query
query = "what is the deadline for admission"
query_embedding = model.encode(query, convert_to_tensor=True)
```

```python
# Compute cosine similarity
scores = cos_sim(query_embedding, question_embeddings)
```

```python
# Get top match
top_idx = scores.argmax()
best_match = df.iloc[top_idx.item()]
```

```python
# Show result
print("User query:", query)
print("Matched question:", best_match["Preprocessed_Question"])
print("Answer:", best_match["Answer"])
```

# EVALUATION #

```python
# Suppose this is your corpus and their embeddings
corpus = ["How to apply?", "Admission deadlines?", "Scholarship opportunities?"]
corpus_embeddings = model.encode(corpus)

# Test set
test_questions = ["When is the last date to apply?", "Tell me about scholarships"]
expected_responses = ["Admission deadlines?", "Scholarship opportunities?"]

correct = 0

for question, expected in zip(test_questions, expected_responses):
    question_embedding = model.encode([question])
    similarities = cosine_similarity(question_embedding, corpus_embeddings)[0]
    best_match_idx = np.argmax(similarities)
    best_match = corpus[best_match_idx]
    
    print(f"User Question: {question}")
    print(f"Expected Match: {expected}")
    print(f"Predicted Match: {best_match}\n")
    
    if best_match == expected:
        correct += 1

accuracy = correct / len(test_questions)
print(f"Accuracy: {accuracy*100:.2f}%")
```

# Saving our Model

```python
class SemanticChatbot:
    def __init__(self, df, index, model):
        self.df = df
        self.index = index
        self.model = model

    def get_response(self, query):
        query_embedding = self.model.encode(query, convert_to_tensor=True)
        question_embeddings = self.model.encode(self.df["Preprocessed_Question"].tolist(), convert_to_tensor=True)
        scores = cos_sim(query_embedding, question_embeddings)
        top_idx = scores.argmax()
        best_match = self.df.iloc[top_idx.item()]
        return best_match["Answer"]
```

```python
chatbot = SemanticChatbot(df=df, index=index, model=model)

# Save using joblib
import joblib
joblib.dump(chatbot, 'chatbot_model.pkl')
```
